```python
import pandas as pd
import os 
import glob
```

```python
dir = '/users/darnej/MPhys/Combined*.csv'

file = glob.glob(dir)[0]

df = pd.read_csv(file)

print(df.head(50))

```

```python
surface_brightness = pd.to_numeric(df['stellar_surf_dens_Msun_kpc2'])
r = pd.to_numeric(df['r_kpc'])
HI_mass = pd.to_numeric(df['MHI_Msun'])



import matplotlib.pyplot as plt


scatter = plt.scatter(r, surface_brightness, c=HI_mass, cmap='viridis', norm='log')
plt.colorbar(scatter, label='HI Mass (Msun)')
plt.xscale('log')
plt.yscale('log')
plt.xlabel('r (kpc)')
plt.ylabel('Stellar Surface Brightness (Msun/kpc^2)')
plt.title('Stellar Surface Brightness vs. Radius')
plt.grid()
plt.show()

```
